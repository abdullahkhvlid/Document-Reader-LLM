```python
import subprocess
import sys

def install_packages():
    subprocess.check_call([sys.executable, "-m", "pip", "install", "torch", "transformers", "PyPDF2", "--quiet"])
    print("Packages installed successfully!")

# Install
install_packages()

# import
import torch
from transformers import pipeline
import PyPDF2
import re
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
class DocumentAIAssistant:
    def __init__(self, model_name="google/flan-t5-base"):
        self.model_name = model_name
        self.document_text = ""
        self.qa_pipeline = None

    def load_document(self, file_path):
        if file_path.endswith('.txt'):
            with open(file_path, 'r', encoding='utf-8') as file:
                self.document_text = file.read()
        elif file_path.endswith('.pdf'):
            text = ""
            with open(file_path, 'rb') as file:
                pdf_reader = PyPDF2.PdfReader(file)
                for page in pdf_reader.pages:
                    text += page.extract_text()
            self.document_text = text

        self.document_text = re.sub(r'\s+', ' ', self.document_text)
        self.document_text = re.sub(r'[^\w\s.,!?-]', '', self.document_text)
        self.document_text = self.document_text.strip()
        return True

    def load_model(self):
        try:
            self.qa_pipeline = pipeline(
                "text2text-generation",
                model=self.model_name,
                device=-1
            )
            return True
        except:
            try:
                self.model_name = "google/flan-t5-small"
                self.qa_pipeline = pipeline(
                    "text2text-generation",
                    model=self.model_name,
                    device=-1
                )
                return True
            except:
                return False

    def answer_question(self, question):
        if not self.document_text or not self.qa_pipeline:
            return "System not ready. Load document and model first."

        prompt = f"""
        Based ONLY on the following document content, answer the question.
        If the answer is not found in the document, say "Answer not found in the document."

        DOCUMENT:
        {self.document_text[:3000]}

        QUESTION: {question}

        ANSWER BASED ONLY ON DOCUMENT:
        """

        try:
            result = self.qa_pipeline(
                prompt,
                max_length=200,
                num_beams=5,
                temperature=0.3,
                repetition_penalty=2.0,
                do_sample=False
            )

            answer = result[0]['generated_text'].strip()

            if "answer not found" in answer.lower() or len(answer) < 5:
                return "Answer not found in the document."

            return answer
        except:
            return "Error generating answer."

print("DocumentAIAssistant class created successfully!")
```
